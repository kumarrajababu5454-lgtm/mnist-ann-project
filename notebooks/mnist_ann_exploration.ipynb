# MNIST ANN exploration

We inspect MNIST before training a fully connected neural network. Images are flattened because a Dense layer expects one feature vector per example.

```python
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.datasets import mnist

(x_train, y_train), (x_test, y_test) = mnist.load_data()
print(x_train.shape, x_test.shape)
print('classes:', np.unique(y_train))
```

MNIST has 60,000 training and 10,000 test examples. Each image is 28 x 28 grayscale pixels, and labels range from 0 to 9.

```python
fig, axes = plt.subplots(1, 10, figsize=(12, 2))
for digit, ax in enumerate(axes):
    ax.imshow(x_train[digit], cmap='gray')
    ax.set_title(str(y_train[digit]))
    ax.axis('off')
plt.show()
```

```python
print('raw pixel range:', x_train.min(), x_train.max())
x_train_normalized = x_train.astype('float32') / 255.0
x_test_normalized = x_test.astype('float32') / 255.0
x_train_flat = x_train_normalized.reshape(len(x_train_normalized), 28 * 28)
print('normalized range:', x_train_normalized.min(), x_train_normalized.max())
print('flattened shape:', x_train_flat.shape)
print('28 x 28 =', 28 * 28)
```

```python
counts = np.bincount(y_train, minlength=10)
plt.bar(range(10), counts)
plt.xlabel('Digit')
plt.ylabel('Training examples')
plt.show()
```
